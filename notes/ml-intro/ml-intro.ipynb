# Intro to machine learning

## 1. Definition

> Machine Learning - Field of study that gives computers the ability to learn
without being explicitly programmed. (Arthur Samuel (1959))

### 1.1 Inductive Learning vs Deductive Learning

- Inductive learning - Specific Examples -> General Rule
- Deductive learning - General Rule -> Specific Examples

So Machine learning is **inductive learning**
