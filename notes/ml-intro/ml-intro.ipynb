# Intro to machine learning

## 1. Definition

> Machine Learning - Field of study that gives computers the ability to learn
without being explicitly programmed. (Arthur Samuel (1959))

### 1.1 Inductive Learning vs Deductive Learning

- Inductive learning - Specific Examples -> General Rule
- Deductive learning - General Rule -> Specific Examples

So Machine learning is **inductive learning**

## 2. Datasets

A machine learning model learns from **data**. Before we talk about the different types of ML,
we need a common vocabulary for describing that data.

### 2.1 Dataset

A **dataset** is a collection of examples that the algorithm learns from.
It is usually represented as a **table**:

- each **row** is one example (an object, an observation)
- each **column** is one property that describes the examples

|        | Humidity | Wind   | Temperature | **Play Tennis** |
|--------|----------|--------|-------------|-----------------|
| **X1** | high     | mid    | 35°         | No              |
| **X2** | mid      | strong | 5°          | No              |
| **X3** | mid      | weak   | 20°         | Yes             |
| **X4** | high     | weak   | 25°         | Yes             |
| **X5** | low      | strong | 18°         | No              |

### 2.2 Instance (Example, Sample, Observation)

An **instance** is a **single row** of the dataset.

- In the table above, `X1`, `X2`, ..., `X5` are instances (each one is a different day).
- Synonyms - *instance*, *example*, *sample*, *observation*, *data point*, *record*.

### 2.3 Attribute (Feature)

An **attribute** is a **column** that describes some property of the instances.
Attributes are the **input** of the model - the information it uses to make a decision.

- In the example: `Humidity`, `Wind` and `Temperature` are attributes.

$$X = (a_1, a_2, \dots, a_N)$$

For example, `X3 = (mid, weak, 20°)`.

The number of attributes `N` is called the **dimensionality** of the data.

*Attribute* and *feature* are used almost interchangeably.
Sometimes "attribute" is the column itself (`Wind`),
and "feature" is the attribute together with its value (`Wind = weak`).

#### Types of attributes

| Type | Description | Example |
|------|-------------|---------|
| **Numerical (continuous)** | Real numbers, we can measure distances and do arithmetic | `Temperature = 20.5°` |
| **Numerical (discrete)** | Whole numbers, usually counts | `Number of rooms = 3` |
| **Categorical (nominal)** | A value from a fixed set, **no order** | `Outlook ∈ {sunny, rainy, overcast}` |
| **Categorical (ordinal)** | A value from a fixed set, **with order** | `Humidity ∈ {low, mid, high}` |
| **Binary** | Only two possible values | `Spam ∈ {yes, no}` |

The type of an attribute matters - many algorithms work only with numbers, so categorical attributes often have to be **encoded** as numbers first (part of *data preprocessing*).

### 2.4 Label (Class, Target)

The **label** (also called **class**, **target** or **output**) is the special column that we want to **predict**. It is the "correct answer" for each instance.

- In the example, the label is `Play Tennis` (Yes / No).
- The label is **not** an input - the model learns to predict it from the attributes.

So the dataset is split into:

$$\underbrace{(a_1, a_2, \dots, a_N)}_{\text{attributes (input) } X} \;\longrightarrow\; \underbrace{y}_{\text{label (output) } Y}$$

- When the label is a **category** (`Yes`/`No`, `cat`/`dog`) → the task is **classification**.
- When the label is a **number** (temperature, price) → the task is **regression**.
- When there is **no label** at all → we can only look for structure in the data (e.g. **clustering**).

A dataset that contains labels is called **labeled**; one without labels is **unlabeled**. Labeling data is often expensive because it usually has to be done by humans.

### 2.5 Prediction

After learning from the dataset, the model receives a **new, unseen instance** - only its attributes, without a label - and has to predict the label:

$$X_t = (low,\ weak,\ 21°) \qquad Class(X_t) = \ ?$$

### 2.6 Training set and Test set

To check whether the model really learned a general rule (and did not just memorize the examples), the dataset is split into two parts:

- **Training set** - the instances the model learns from (usually ~70-80%).
- **Test set** - instances the model **never sees** during training;
used to measure how well it works on new data (~20-30%).
- **Validation set** - used to tune the model while training,
so that the test set stays completely untouched until the final evaluation.

### 3.3 Clustering

The goal is to split the instances into groups (**clusters**) so that:

- instances in the **same** cluster are **similar** to each other
- instances in **different** clusters are **dissimilar**

The number of clusters and their meaning are **not** given in advance - the algorithm has to
discover the structure in the data by itself.

[Clustering example](../../official-materials/ml-intro.pdf#page=65)

Similarity is usually measured with a **distance** between instances (e.g. Euclidean distance) - the smaller
the distance, the more similar the instances.

To decide which two clusters are the most similar, we need a distance between **clusters** ([Cluster Similarity](../../official-materials/ml-intro.pdf#page=68)):

|          Linkage          |                      Distance between clusters $C_1$ and $C_2$                       |                                                Formula                                                |
|:-------------------------:|:------------------------------------------------------------------------------------:|:-----------------------------------------------------------------------------------------------------:|
|    **Single Linkage**     |                distance of the two **closest** members (most similar)                |                            $\min\limits_{x \in C_1,\, y \in C_2} d(x, y)$                             |
|   **Complete Linkage**    |               distance of the two **farthest** members (least similar)               |                            $\max\limits_{x \in C_1,\, y \in C_2} d(x, y)$                             |
| **Group Average Linkage** |                  **average** distance between all pairs of members                   | $\frac{1}{\lvert C_1\rvert \lvert C_2\rvert} \sum\limits_{x \in C_1} \sum\limits_{y \in C_2} d(x, y)$ |
|   **Centroid Linkage**    | distance between the **centroids** (mean points) $\mu_1$ and $\mu_2$ of the clusters |                                           $d(\mu_1, \mu_2)$                                           |

Hierarchical clustering builds a **hierarchy (tree) of clusters**. The tree is called a **dendrogram**:

- the **leaves** are the individual instances
- every inner node is a cluster made by joining its children
- the **height** at which two clusters are joined shows how (dis)similar they are - clusters joined low in the tree are very similar, clusters joined near the root are very different

There are two approaches for ([Hierarchical Clustering](../../official-materials/ml-intro.pdf#page=67)):

|     Approach      | Direction |               Start               |                Each step                |                End                |
|:-----------------:|:---------:|:---------------------------------:|:---------------------------------------:|:---------------------------------:|
| **Agglomerative** | bottom-up | every instance is its own cluster | **merge** the two most similar clusters |  one cluster with all instances   |
|   **Divisive**    | top-down  |  one cluster with all instances   |      **split** a cluster into two       | every instance is its own cluster |


### 3.5 Reinforcement Learning

In **reinforcement learning** there is no dataset with correct answers.
Instead, an **agent** learns by **trial and error**:
it acts in an **environment** and gets a **reward** (good) or a **penalty** (bad) for what it does.

The goal of the agent is to learn which actions bring the **most reward in the long run**.

|      Term       |                                    Meaning                                    |           Example (a game)            |
|:---------------:|:-----------------------------------------------------------------------------:|:-------------------------------------:|
|    **Agent**    |                         the learner / decision maker                          |              the player               |
| **Environment** |                          the world the agent acts in                          |               the game                |
|    **State**    |                   the current situation of the environment                    |          the current screen           |
|   **Action**    |                             what the agent can do                             |        move left, right, jump         |
|   **Reward**    |      feedback after an action - positive (reward) or negative (penalty)       |      +1 for a coin, -1 for dying      |
|   **Policy**    |       the strategy the agent learns: which action to take in each state       |     "jump when an enemy is close"     |

The learning loop:

1. The agent sees the current **state**.
2. It chooses an **action**.
3. The environment changes to a new **state** and gives a **reward**.
4. The agent updates its **policy** so that good actions become more likely.
5. Repeat.
