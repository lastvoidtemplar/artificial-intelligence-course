## 1. Local Search

### 1.1 Review

**Local Search** strategies only look at the local area, so they can only look at states in this area.
If the solution is outside of it, they will not be able to find it.

- **Local Search** in Artificial Intelligence is an optimizing algorithm to find the optimal solution more quickly.
- **Local search algorithms** are used when we care only about a solution but not the path to a solution.
- **Local search** is a heuristic method for solving computationally hard optimization problems.
- **Local search** can be used on problems that can be formulated as finding a solution maximizing a criterion among a number of candidate solutions.
- **Local search algorithms** move from solution to solution in the space of candidate solutions (the search space) by
applying local changes, until a solution deemed optimal is found or a time bound has elapsed.

## 2. Constraint Satisfaction Problems (CSPs)

### 2.1 Definitions

In standard search problems, the **state** is a *black box*, which supports only 3 functions:

- *successor function* - ```Fn(State) -> [(Action, State)]```
- *goal test* - ```Fn(State) -> bool```
- *eval* (f(n)) - ```Fn(State) -> R```
    - In Uniform-Cost Search - f(n) = g(n)
    - In Greedy Best-First Search - f(n) = h(n)
    - In A* - f(n) = g(n) + h(n)

🔑 In a CSP

- **state** is defined by a set of <ins>variables</ins> **X<sub>i</sub>** with <ins>values</ins> from <ins>domain</ins> **D<sub>i</sub>**
- **goal test** is a set of <ins>constraints</ins> specifying allowable combinations of values of subsets of variables

### 2.3 Varieties

Variables can be discrete or continuous.

Discrete variables can have finite or infinite domains.

If we have **n** discrete variables with domains that have size ≤ **d**,
then the number of possible complete assignments is O(d<sup>n</sup>)

Example of a CSP with discrete variables with infinite domains
is job scheduling, where variables are start/finish days for each job
(**StartJob<sub>1</sub> + 5 ≤ StartJob<sub>3</sub>**).
If constraints are linear, the problem is *solvable* (there is an algorithm that always stops and correctly says "here's a solution" or "no solution exists").
Integer linear feasibility is NP-complete. If constraints are nonlinear, then the problem is *undecidable* (no general algorithm exists at all).

If the problem has continuous variables and linear constraints, the problem is solvable in poly time (O(n<sup>k</sup>)).



Constraints can be unary, binary, or higher-order.

Unary constraints involve a single variable. For example, **SA ≠ green**

Binary constraints involve pairs of variables. For example, **SA ≠ WA**

Higher-order constraints involve 3 or more variables. For example, **X<sub>1</sub> + X<sub>2</sub> ≤ X<sub>3</sub>**

Constraints can be soft (so-called preferences), which are often represented by a cost for each variable assignment.
This leads to *constraint optimization problems*. For example, **red** is better than **green**.

## 3. Homework - N-Queens

### 3.1 Problem Summary

Place N chess queens on an NxN chessboard so that no two queens threaten each other.
In other words, the solution requires that no two queens share the same row, column, or diagonal.

Use the MinConflicts algorithm to solve the problem.

### 3.2 Naive Problem Formulation

N*N **variables** with domain **{ empty, queen }**

3 types of **constraints**:

- One queen per column
- One queen per row
- At most one queen per diagonal

### 3.3 Min-Conflicts

**Local Search** algorithms typically work with **complete** states (all variables are assigned).
To apply these algorithms to CSPs, we must allow states with unsatisfied constraints and operators to **reassign** variable values.

🔑 How do we select a variable? - Randomly select any conflicting variable

🔑 How do we select a value? - min-conflicts heuristic (the value that violates the **fewest** constraints).

Min-Conflicts = Hill Climbing with h(n) = total number of violated constraints.

Given a random initial state, it can solve n-queens in almost constant time for arbitrary n with high probability (e.g., n = 10,000,000).
The same appears to be true for any randomly-generated CSP except in a narrow range of the ratio.

### 3.6 Max-Min-Optimization

In Min-Conflicts we select conflicting variables randomly.
Wouldn't it be better to choose the variables which have the highest number of conflicts?
Yes, but asterisk.

🔑 If we are not careful, **LOOPS** are possible! [4-Queens](../../official-materials/csp.pdf#page=58)

So what can we do? The answer is a **randomization** when there is a tie!

🔑 How do we select a variable? - The variable with a current value that has the highest number of conflicts.
If there are multiple, select **RANDOMLY**.

🔑 How do we select a value? - min-conflicts heuristic (the value that violates the **fewest** constraints).
If there are multiple, select **RANDOMLY**.
